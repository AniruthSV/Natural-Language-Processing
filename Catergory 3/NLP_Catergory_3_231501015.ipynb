```python
import nltk
import re
from nltk.corpus import reuters

nltk.download('reuters')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
def preprocess(text):

    text = text.lower()

    tokens = nltk.word_tokenize(text)

    tokens = [w for w in tokens if w.isalpha()]

    return tokens
```

```python
def sc_rule_based_morphology(word):
    suffix_rules = ['ing','ed','ly','es','s','er','est']
    for suffix in suffix_rules:
        if word.endswith(suffix) and len(word) > len(suffix)+2:
            root = word[:-len(suffix)]
            return root, suffix
    return word, "None"
```

```python
small_corpus = """
The boys were playing happily in the gardens.
She quickly finished her work and walked away.
The bigger dogs chased the smaller cats.
"""
```

```python
tokens_small = preprocess(small_corpus)
print("Morphological Analysis for Small Corpus\n")
for word in tokens_small:
    root, suffix = sc_rule_based_morphology(word)
    print(word, " -> Root:", root, "| Suffix:", suffix)
```

```python
def reu_rule_based_morphology(word):

    suffix_rules = [
        "ization","ation","ition",
        "ingly","edly",
        "ing","ed","en",
        "ies","es","s",
        "ly",
        "er","est",
        "ment","ness","ship",
        "tion","sion",
        "able","ible",
        "al","ous",
        "ist","ism",
        "ity"
    ]

    prefix_rules = [
        "un","re","dis","pre","mis","over","under"
    ]


    root = word
    suffix = None
    prefix = None


    # PREFIX DETECTION
    for p in prefix_rules:
        if word.startswith(p) and len(word) > len(p)+2:
            prefix = p
            root = word[len(p):]
            break


    # SUFFIX DETECTION
    for s in suffix_rules:

        if root.endswith(s) and len(root) > len(s)+2:

            base = root[:-len(s)]

            # spelling correction rules
            if s == "ies":
                base = base + "y"

            if s == "ing" and base.endswith(base[-1]):
                base = base[:-1]

            root = base
            suffix = s
            break

    return prefix, root, suffix
```

```python
text = ""

for file in reuters.fileids()[:20]:
    text += reuters.raw(file)

tokens = preprocess(text)
```

```python
print("Morphological Analysis on Reuters Corpus\n")

for word in tokens[:200]:

    prefix, root, suffix = reu_rule_based_morphology(word)

    print(f"{word} -> Prefix:{prefix} | Root:{root} | Suffix:{suffix}")
```
