```python
import nltk
import re
from collections import Counter
import math
nltk.download('brown')
from nltk.corpus import brown
```

```python
def preprocess(text):

    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()

    return tokens
```

```python
def build_unigram(tokens):

    word_counts = Counter(tokens)

    total_words = len(tokens)

    vocabulary = set(tokens)

    vocab_size = len(vocabulary)

    return word_counts, total_words, vocab_size
```

```python
def unigram_probability(word, word_counts, total_words):

    return word_counts[word] / total_words
```

```python
def laplace_probability(word, word_counts, total_words, vocab_size):

    return (word_counts[word] + 1) / (total_words + vocab_size)
```

```python
def sentence_probability(sentence, word_counts, total_words, vocab_size):

    tokens = preprocess(sentence)

    prob = 1

    for word in tokens:

        prob *= laplace_probability(word, word_counts, total_words, vocab_size)

    return prob
```

```python
print("\n----- SMALL CORPUS -----\n")

small_corpus = """
the cat sat on the mat
the dog sat on the log
the cat chased the mouse
"""

tokens_small = preprocess(small_corpus)

word_counts_small, total_small, vocab_small = build_unigram(tokens_small)

print("Vocabulary Size:", vocab_small)
print("Total Words:", total_small)

print("\nWord Probabilities (Laplace):\n")

for word in word_counts_small:

    p = laplace_probability(word, word_counts_small, total_small, vocab_small)

    print(word, ":", round(p,5))


test_sentence = "the cat sat"

prob = sentence_probability(test_sentence, word_counts_small, total_small, vocab_small)

print("\nSentence Probability:", prob)
```

```python
print("\n----- BROWN CORPUS -----\n")

brown_words = brown.words()[:50000]  # using subset for speed

brown_text = " ".join(brown_words)

tokens_brown = preprocess(brown_text)

word_counts_brown, total_brown, vocab_brown = build_unigram(tokens_brown)

print("Vocabulary Size:", vocab_brown)
print("Total Words:", total_brown)


# Sample probabilities

sample_words = ["the","government","science","computer"]

print("\nSample Word Probabilities (Laplace):\n")

for word in sample_words:

    p = laplace_probability(word, word_counts_brown, total_brown, vocab_brown)

    print(word, ":", p)


test_sentence = "the government is working"

prob = sentence_probability(test_sentence, word_counts_brown, total_brown, vocab_brown)

print("\nSentence Probability:", prob)
```
