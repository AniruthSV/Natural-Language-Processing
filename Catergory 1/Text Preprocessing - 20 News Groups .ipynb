```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.datasets import fetch_20newsgroups
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
newsgroups = fetch_20newsgroups(subset='train',
                                categories=['sci.med'],
                                remove=('headers', 'footers', 'quotes'))
raw_text = newsgroups.data[0]
print(f"--- 1. RAW TEXT (Medical Post) ---\n{raw_text[:150]}...\n")
```

```python
#Normalization
normalized_text = raw_text.lower()
```

```python
#Tokenization
tokens = word_tokenize(normalized_text)
print(f"--- 2. TOKENS (First 10) ---\n{tokens[:10]}\n")
```

```python
#Stop Word Removal
stop_words = set(stopwords.words('english'))
punctuation = set(string.punctuation)
filtered_tokens = [w for w in tokens if w not in stop_words and w not in punctuation and w.isalpha()]
print(f"--- 3. FILTERED TOKENS ---\n{filtered_tokens[:10]}\n")
```

```python
#Stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(w) for w in filtered_tokens]
print(f"--- 4. STEMMED TOKENS ---\n{stemmed_tokens[:10]}\n")
```

```python
#Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(w) for w in filtered_tokens]
print(f"--- 5. LEMMATIZED TOKENS ---\n{lemmatized_tokens[:10]}\n")
```
