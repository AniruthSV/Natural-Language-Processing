```python
import nltk
import re
from collections import defaultdict
from nltk.corpus import treebank

nltk.download('treebank')
nltk.download('universal_treebanks')
```

```python
class SC_FSAMorphAnalyzer:

    def __init__(self):

        # Prefixes
        self.prefixes = ['un', 're', 'in', 'dis']

        # Suffixes with labels
        self.suffixes = {
            'ing': 'Present Participle',
            'ed': 'Past Tense',
            's': 'Plural',
            'ly': 'Adverb',
            'er': 'Comparative'
        }

    def sc_analyze(self, word):

        prefix = None
        suffix = None
        root = word

        # PREFIX TRANSITION
        for p in self.prefixes:
            if word.startswith(p) and len(word) > len(p)+2:
                prefix = p
                root = word[len(p):]
                break

        # SUFFIX TRANSITION
        for s in self.suffixes:
            if root.endswith(s) and len(root) > len(s)+2:
                suffix = s
                root = root[:-len(s)]
                break

        return {
            "word": word,
            "prefix": prefix,
            "root": root,
            "suffix": suffix,
            "suffix_type": self.suffixes.get(suffix, None)
        }
```

```python
manual_corpus = [
    "running", "played", "cats", "dogs",
    "unhappy", "redo", "dislike",
    "quickly", "bigger", "reading"
]

sc_analyzer = SC_FSAMorphAnalyzer()

print("\nMorphological Analysis (Manual Corpus)\n")

for word in manual_corpus:
    print(sc_analyzer.sc_analyze(word))
```

```python
class FSAMorphAnalyzer:

    def __init__(self):

        # Expanded Prefix List
        self.prefixes = [
            "un","re","dis","in","im","non",
            "over","under","pre","post","mis"
        ]

        # Expanded Suffix Rules
        self.suffixes = {
            "ing":"Present Participle",
            "ed":"Past Tense",
            "s":"Plural",
            "es":"Plural",
            "en":"Past Participle",
            "ly":"Adverb",
            "ness":"Noun Formation",
            "ment":"Noun Formation",
            "er":"Agent/Comparative",
            "est":"Superlative",
            "able":"Adjective",
            "ful":"Adjective",
            "less":"Adjective",
            "tion":"Noun",
            "ity":"Noun",
            "al":"Adjective"
        }

    def spelling_rules(self, root):

        # double consonant rule
        if len(root) > 2 and root[-1] == root[-2]:
            root = root[:-1]

        # y -> i rule
        if root.endswith("i"):
            root = root[:-1] + "y"

        return root


    def analyze(self, word):

        original = word
        prefix = None
        suffix = None
        root = word

        # PREFIX CHECK
        for p in self.prefixes:
            if word.startswith(p) and len(word) > len(p)+2:
                prefix = p
                root = word[len(p):]
                break

        # SUFFIX CHECK
        for s in sorted(self.suffixes, key=len, reverse=True):

            if root.endswith(s) and len(root) > len(s)+2:
                suffix = s
                root = root[:-len(s)]
                break

        # Apply spelling normalization
        root = self.spelling_rules(root)

        return {
            "word": original,
            "prefix": prefix,
            "root": root,
            "suffix": suffix,
            "suffix_type": self.suffixes.get(suffix)
        }
```

```python
sentences = treebank.sents()

words = []

for sent in sentences[:200]:
    for w in sent:
        if w.isalpha():
            words.append(w.lower())

analyzer = FSAMorphAnalyzer()

print("\nUD Treebank Analysis\n")

for word in words[:50]:
    # Note: 'analyzer' is currently an instance of SC_FSAMorphAnalyzer (from cell ZITgzGzQJls1),
    # which has a method 'sc_analyze', not 'analyze'. This line may cause an AttributeError
    # after the NameError is resolved, unless 'analyzer' is re-instantiated as FSAMorphAnalyzer.
    print(analyzer.analyze(word))
```
